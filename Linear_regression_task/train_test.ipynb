```python
# Parameters
d = 3
sigma = 0.01
n_train = 100
n_test = 1000
beta = np.random.randn(d)
```

```python
mlp_model, loss_history= train_flexible_mlp(
    X_train, 
    y_train, 
    epochs=1000, 
    lr=0.01, 
    hidden_layers=[2048], 
    activation=nn.ReLU
)
```

```python
mlp_model_cos, loss_history_cos = train_flexible_mlp(
    X_train, 
    y_train, 
    epochs=10000, 
    lr=0.01, 
    hidden_layers=[2048], 
    activation=CosActivation
)
```

```python
test_error = test_model(mlp_model, d, sigma, n_test, beta)
print(f"Test MSE with custom MLP (ReLU): {test_error}")

test_error_cos = test_model(mlp_model_cos, d, sigma, n_test, beta)
print(f"Test MSE with custom MLP (Cos): {test_error_cos}")
```
