```python
for i in range(60, 80):
    folder='uniformNorms/run_' + str(i)
    os.makedirs("weights/" + folder + "/", exist_ok=True)
    os.makedirs("outputs/" + folder + "/", exist_ok=True)

    losses, weights, norms, r = collect_clusterPredictions(hidden_dim, center_list, var_list, target_fn="uniformMap", opt='Adam', lr=0.00005, max_epochs=40000)
    print(r)

    plot_list = range(len(loss_n))
    plot_trainingLoss(np.mean(losses[0], axis=0, keepdims=True), loss_n, plot_list, hidden_dim[0], epochs=40000, folder=folder, file="trainAvg40000")

    for j in range(k):
        for layer in range(2):
            torch.save({'linear.weight': weights[0][j][2*layer], 'linear.bias': weights[0][j][2*layer+1]}, 
                "weights/" + folder + "/layer=%s" % layer + "_k="+str(j)+".pth")
            
    with open("outputs/" + folder + "/finalTrain.txt", "w") as file:
        for j in range(k):
            file.write("r = " + str(r) + ": " + str(losses[0][j][-1]) + "\n")

    with open("outputs/" + folder + "/norms.txt", "w") as file:
        for j in range(k):
            file.write(str(norms[0][j]) + "\n")
        
```
